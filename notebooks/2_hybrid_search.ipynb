```python
import sys
from pathlib import Path
sys.path.append(str(Path("..").resolve()))
```

```python
from src.data_utils import load_chunks
from src.retrievers import create_bm25,create_hybrid_retriever
from src.loaders import load_vectorstore
```

```python
#chunks loading from saved data
chunks=load_chunks()

vectorstore=load_vectorstore()

bm25=create_bm25(chunks)

hybrid_retriever=create_hybrid_retriever(bm25,vectorstore)
```

```python
# testing 
query="what is the attention mechanism?"

bm25_results=bm25.invoke(query)

vector_results=vectorstore.as_retriever().invoke(query)


# hybrid resutls
hybrid_results=hybrid_retriever.invoke(query)

# 3. Display Comparison
print(f"Query: {query}\n" + "="*50)

def print_results(title, results):
    print(f"\n--- {title} ---")
    for i, doc in enumerate(results[:3]): # Show top 3 for comparison
        print(f"{i+1}: {doc.page_content}...")

print_results("BM25 (Keyword Search)", bm25_results)
print_results("Vector (Semantic Search)", vector_results)
print_results("Hybrid Search (Combined)", hybrid_results)
```
